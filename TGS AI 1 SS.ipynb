## Metode Ground Penetrating Radar (GPR)

1. **Role:** Anda berperan sebagai seorang ahli geofisika yang memiliki keahlian dalam metode Ground Penetrating Radar (GPR), khususnya dalam pengolahan dan interpretasi radargram. Anda mampu menganalisis waktu tempuh gelombang elektromagnetik, kedalaman objek, refleksi bawah permukaan, serta mengidentifikasi perubahan karakteristik material berdasarkan pola refleksi pada radargram.

   **Output:** Menghasilkan analisis data GPR yang sistematis berupa hasil pengolahan radargram, estimasi kedalaman, identifikasi pola refleksi, serta interpretasi kondisi bawah permukaan.



```python

import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# ANALISIS GROUND PENETRATING RADAR (GPR)
# Simulasi Radargram dan Interpretasi
# ==========================================

# 1. Parameter simulasi
np.random.seed(42)

jumlah_trace = 200
jumlah_sampel = 300

waktu_maksimum = 100  # ns
kecepatan = 0.10      # m/ns

# Membuat sumbu waktu
waktu = np.linspace(0, waktu_maksimum, jumlah_sampel)

# Membuat sumbu posisi lintasan
posisi = np.linspace(0, 10, jumlah_trace)

# 2. Membuat data radargram sintetis
radargram = np.zeros((jumlah_sampel, jumlah_trace))

# Refleksi lapisan bawah permukaan
for i in range(jumlah_trace):
    # Refleksi lapisan pertama
    t1 = 20 + 2 * np.sin(posisi[i])

    # Refleksi lapisan kedua
    t2 = 45 + 3 * np.sin(posisi[i] * 0.7)

    # Simulasi refleksi objek berbentuk hiperbola
    x_objek = 5.0
    t_objek = 35 + 8 * np.sqrt(
        1 + ((posisi[i] - x_objek) / 1.2) ** 2
    )

    # Membentuk sinyal refleksi
    radargram[:, i] += np.exp(
        -((waktu - t1) / 1.5) ** 2
    )

    radargram[:, i] += 0.7 * np.exp(
        -((waktu - t2) / 2.0) ** 2
    )

    radargram[:, i] += 1.2 * np.exp(
        -((waktu - t_objek) / 1.5) ** 2
    )

# Menambahkan noise
radargram += 0.15 * np.random.randn(
    jumlah_sampel, jumlah_trace
)

# 3. Normalisasi amplitudo
radargram = radargram / np.max(np.abs(radargram))

# 4. Konversi waktu menjadi kedalaman
kedalaman = waktu * kecepatan / 2

# 5. Visualisasi radargram
plt.figure(figsize=(12, 6))

plt.imshow(
    radargram,
    cmap="seismic",
    aspect="auto",
    extent=[
        posisi.min(),
        posisi.max(),
        kedalaman.max(),
        kedalaman.min()
    ],
    vmin=-1,
    vmax=1
)

plt.xlabel("Posisi Lintasan (m)")
plt.ylabel("Kedalaman (m)")
plt.title("Simulasi Radargram Ground Penetrating Radar")
plt.colorbar(label="Amplitudo Relatif")
plt.tight_layout()
plt.show()

# 6. Analisis amplitudo rata-rata
amplitudo_rata = np.mean(
    np.abs(radargram), axis=1
)

plt.figure(figsize=(8, 5))

plt.plot(amplitudo_rata, kedalaman)
plt.gca().invert_yaxis()

plt.xlabel("Rata-rata Amplitudo Absolut")
plt.ylabel("Kedalaman (m)")
plt.title("Analisis Amplitudo terhadap Kedalaman")
plt.grid(True)
plt.tight_layout()
plt.show()

# 7. Menampilkan informasi hasil simulasi
print("=== HASIL SIMULASI GPR ===")
print(f"Jumlah trace       : {jumlah_trace}")
print(f"Jumlah sampel      : {jumlah_sampel}")
print(f"Waktu maksimum     : {waktu_maksimum} ns")
print(f"Kecepatan gelombang: {kecepatan} m/ns")
print(f"Kedalaman maksimum : {kedalaman.max():.2f} m")

print("\nInterpretasi awal:")
print("- Refleksi menunjukkan batas atau perubahan sifat material.")
print("- Pola melengkung dapat menyerupai refleksi hiperbolik objek.")
print("- Noise merupakan gangguan yang ditambahkan secara simulasi.")
print("- Data ini sintetis, bukan hasil pengukuran lapangan.")

```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =====================================================
# TASK 5: ANALISIS GPR BERDASARKAN PRINSIP FISIKA
# =====================================================

# 1. PARAMETER SURVEI
panjang_lintasan = 20.0       # meter
waktu_maksimum_ns = 100.0     # nanosekon
kecepatan_m_per_ns = 0.10     # meter/ns
frekuensi_MHz = 250.0         # MHz

jumlah_trace = 100
jumlah_sampel = 300

# Data belum berasal dari survei asli
status_data = "SIMULASI - bukan data lapangan"

# 2. MEMBUAT POSISI DAN WAKTU
posisi = np.linspace(
    0, panjang_lintasan, jumlah_trace
)

waktu_ns = np.linspace(
    0, waktu_maksimum_ns, jumlah_sampel
)

# 3. MEMBUAT RADARGRAM SINTETIS
np.random.seed(42)

radargram = np.zeros(
    (jumlah_sampel, jumlah_trace)
)

for i, x in enumerate(posisi):

    # Reflektor sintetis pertama
    t1 = 20 + 1.5 * np.sin(x / 3)

    # Reflektor sintetis kedua
    t2 = 45 + 2 * np.sin(x / 4)

    # Respons hiperbolik sintetis
    t3 = 35 + 8 * np.sqrt(
        1 + ((x - 10) / 1.5) ** 2
    )

    radargram[:, i] = (
        np.exp(-((waktu_ns - t1) / 1.5) ** 2)
        - 0.7 * np.exp(-((waktu_ns - t2) / 2) ** 2)
        + 1.2 * np.exp(-((waktu_ns - t3) / 1.8) ** 2)
        + 0.08 * np.random.randn(jumlah_sampel)
    )

# Normalisasi amplitudo
radargram /= np.max(np.abs(radargram))

# 4. KONVERSI WAKTU TEMPUH MENJADI KEDALAMAN
# TWT adalah waktu perjalanan pergi-pulang.
# v dalam m/ns dan TWT dalam ns menghasilkan kedalaman dalam meter.

kedalaman_m = (
    kecepatan_m_per_ns * waktu_ns / 2
)

# 5. MENGANALISIS PUNCAK AMPLITUDO SETIAP TRACE
hasil = []

for i, x in enumerate(posisi):

    trace = radargram[:, i]

    indeks = np.argmax(np.abs(trace))

    twt = waktu_ns[indeks]
    kedalaman = kecepatan_m_per_ns * twt / 2
    amplitudo = trace[indeks]

    # Kategori hanya berdasarkan besar amplitudo relatif.
    # Bukan penentuan jenis material.
    if abs(amplitudo) >= 0.7:
        interpretasi = "Respons refleksi relatif kuat"
    elif abs(amplitudo) >= 0.3:
        interpretasi = "Respons refleksi relatif sedang"
    else:
        interpretasi = "Respons refleksi relatif lemah"

    hasil.append({
        "Posisi_m": x,
        "TWT_ns": twt,
        "Kecepatan_m_per_ns": kecepatan_m_per_ns,
        "Kedalaman_m": kedalaman,
        "Amplitudo": amplitudo,
        "Karakteristik": interpretasi
    })

df = pd.DataFrame(hasil)

# 6. VALIDASI PARAMETER
if kecepatan_m_per_ns <= 0:
    raise ValueError("Kecepatan gelombang harus positif.")

if waktu_maksimum_ns <= 0 or panjang_lintasan <= 0:
    raise ValueError("Waktu dan panjang lintasan harus positif.")

print("===== PARAMETER ANALISIS GPR =====")
print(f"Status data        : {status_data}")
print(f"Panjang lintasan   : {panjang_lintasan:.2f} m")
print(f"Waktu maksimum     : {waktu_maksimum_ns:.2f} ns")
print(f"Kecepatan gelombang: {kecepatan_m_per_ns:.3f} m/ns")
print(f"Frekuensi antena   : {frekuensi_MHz:.1f} MHz")
print(f"Kedalaman maksimum : {kedalaman_m.max():.2f} m")

print("\n===== CONTOH HASIL PENGOLAHAN =====")
print(df.head(10).to_string(index=False))

# 7. SIMPAN HASIL KE CSV
df.to_csv("hasil_analisis_gpr.csv", index=False)
print("\nData tersimpan: hasil_analisis_gpr.csv")

# 8. VISUALISASI RADARGRAM TERHADAP WAKTU
plt.figure(figsize=(12, 6))

plt.imshow(
    radargram,
    cmap="gray",
    aspect="auto",
    extent=[
        posisi.min(),
        posisi.max(),
        waktu_maksimum_ns,
        0
    ]
)

plt.xlabel("Posisi Lintasan (m)")
plt.ylabel("Waktu Tempuh Dua Arah (ns)")
plt.title("Radargram GPR - Data Simulasi")
plt.colorbar(label="Amplitudo Relatif")
plt.tight_layout()
plt.show()

# 9. VISUALISASI PENAMPANG KEDALAMAN
plt.figure(figsize=(12, 6))

plt.imshow(
    radargram,
    cmap="seismic",
    aspect="auto",
    extent=[
        posisi.min(),
        posisi.max(),
        kedalaman_m.max(),
        kedalaman_m.min()
    ],
    vmin=-1,
    vmax=1
)

plt.xlabel("Posisi Lintasan (m)")
plt.ylabel("Estimasi Kedalaman (m)")
plt.title("Penampang GPR Berdasarkan Kecepatan Asumsi")
plt.colorbar(label="Amplitudo Relatif")
plt.tight_layout()
plt.show()

# 10. BATASAN INTERPRETASI
print("\n===== CATATAN INTERPRETASI =====")
print("1. Kedalaman dihitung dari d = v x TWT / 2.")
print("2. Kecepatan harus mewakili material bawah permukaan.")
print("3. Amplitudo menunjukkan respons refleksi, bukan jenis material.")
print("4. Reflektor dan hiperbola perlu diperiksa pada radargram.")
print("5. Penentuan tanah, batuan, rongga, atau rekahan")
print("   memerlukan bukti pendukung dan validasi geologi.")
print("6. Hasil ini simulasi, bukan interpretasi data lapangan.")

```

## Metode Mikrotremor HVSR

1. **Role:** Anda berperan sebagai seorang ahli geofisika yang memiliki keahlian dalam metode mikrotremor, khususnya Horizontal to Vertical Spectral Ratio (HVSR). Anda mampu mengolah data getaran tanah dalam tiga komponen, yaitu dua komponen horizontal dan satu komponen vertikal, menghitung rasio spektral H/V, menentukan frekuensi dominan, serta menginterpretasikan karakteristik lapisan bawah permukaan.

   **Output:** Menghasilkan analisis mikrotremor berupa kurva HVSR, frekuensi dominan, nilai amplifikasi, serta interpretasi karakteristik lapisan bawah permukaan.

```python

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.signal import welch, find_peaks

# =====================================================
# ANALISIS MIKROTREMOR METODE HVSR
# Horizontal to Vertical Spectral Ratio
# =====================================================

# 1. PARAMETER PENGUKURAN
np.random.seed(42)

nama_lokasi = "Stasiun Mikrotremor 01"
durasi = 300            # detik
sampling_rate = 100     # Hz
jumlah_data = durasi * sampling_rate

# Rentang frekuensi analisis
f_min = 0.2
f_max = 20.0

# 2. SIMULASI DATA TIGA KOMPONEN
# Ganti bagian ini dengan data lapangan jika tersedia.

waktu = np.arange(jumlah_data) / sampling_rate

# Komponen horizontal Utara-Selatan (NS)
ns = (
    0.8 * np.sin(2 * np.pi * 1.5 * waktu)
    + 0.3 * np.sin(2 * np.pi * 4.0 * waktu)
    + 0.4 * np.random.randn(jumlah_data)
)

# Komponen horizontal Timur-Barat (EW)
ew = (
    0.7 * np.sin(2 * np.pi * 1.5 * waktu + 0.5)
    + 0.25 * np.sin(2 * np.pi * 3.5 * waktu)
    + 0.4 * np.random.randn(jumlah_data)
)

# Komponen vertikal (Z)
z = (
    0.35 * np.sin(2 * np.pi * 1.5 * waktu)
    + 0.2 * np.sin(2 * np.pi * 5.0 * waktu)
    + 0.4 * np.random.randn(jumlah_data)
)

# 3. MENGHITUNG SPEKTRUM FOURIER
frekuensi, psd_ns = welch(
    ns, fs=sampling_rate, nperseg=4096
)

_, psd_ew = welch(
    ew, fs=sampling_rate, nperseg=4096
)

_, psd_z = welch(
    z, fs=sampling_rate, nperseg=4096
)

# Amplitudo spektral dari PSD
amp_ns = np.sqrt(psd_ns)
amp_ew = np.sqrt(psd_ew)
amp_z = np.sqrt(psd_z)

# 4. KOMPONEN HORIZONTAL GABUNGAN
# Geometric mean horizontal
amp_h = np.sqrt(amp_ns * amp_ew)

# 5. MENGHITUNG RASIO HVSR
hvsr = amp_h / np.maximum(amp_z, 1e-12)

# Memilih rentang frekuensi analisis
mask = (
    (frekuensi >= f_min)
    & (frekuensi <= f_max)
)

freq_analisis = frekuensi[mask]
hvsr_analisis = hvsr[mask]

# 6. MENENTUKAN FREKUENSI DOMINAN
puncak, properti = find_peaks(hvsr_analisis)

if len(puncak) > 0:
    indeks_puncak = puncak[
        np.argmax(hvsr_analisis[puncak])
    ]
else:
    indeks_puncak = np.argmax(hvsr_analisis)

f0 = freq_analisis[indeks_puncak]
amplifikasi = hvsr_analisis[indeks_puncak]

# 7. INTERPRETASI AWAL FREKUENSI DOMINAN
if f0 < 1:
    interpretasi = (
        "Frekuensi dominan rendah; dapat berkaitan "
        "dengan lapisan sedimen yang relatif tebal."
    )
elif f0 < 5:
    interpretasi = (
        "Frekuensi dominan menengah; kemungkinan "
        "menunjukkan respons lapisan tanah atau sedimen."
    )
else:
    interpretasi = (
        "Frekuensi dominan tinggi; dapat berkaitan "
        "dengan lapisan permukaan yang relatif tipis "
        "atau material yang lebih kaku."
    )

# 8. MENAMPILKAN HASIL
print("=" * 50)
print("HASIL ANALISIS MIKROTREMOR - METODE HVSR")
print("=" * 50)

print(f"Lokasi pengukuran       : {nama_lokasi}")
print(f"Durasi rekaman          : {durasi} detik")
print(f"Sampling rate           : {sampling_rate} Hz")
print(f"Frekuensi dominan (f0)  : {f0:.3f} Hz")
print(f"Amplitudo puncak HVSR   : {amplifikasi:.3f}")
print(f"Interpretasi awal       : {interpretasi}")

# 9. MENYIMPAN DATA HVSR KE CSV
df = pd.DataFrame({
    "Frekuensi_Hz": freq_analisis,
    "HVSR": hvsr_analisis
})

df.to_csv("hasil_analisis_HVSR.csv", index=False)

print("\nData kurva HVSR tersimpan sebagai:")
print("hasil_analisis_HVSR.csv")

# 10. GRAFIK TIGA KOMPONEN DALAM DOMAIN WAKTU
plt.figure(figsize=(12, 7))

plt.subplot(3, 1, 1)
plt.plot(waktu[:2000], ns[:2000])
plt.ylabel("NS")
plt.title("Sinyal Mikrotremor Tiga Komponen")

plt.subplot(3, 1, 2)
plt.plot(waktu[:2000], ew[:2000])
plt.ylabel("EW")

plt.subplot(3, 1, 3)
plt.plot(waktu[:2000], z[:2000])
plt.ylabel("Z")
plt.xlabel("Waktu (detik)")

plt.tight_layout()
plt.show()

# 11. GRAFIK KURVA HVSR
plt.figure(figsize=(10, 6))

plt.plot(
    freq_analisis,
    hvsr_analisis,
    color="blue",
    linewidth=1.5,
    label="Kurva HVSR"
)

plt.axvline(
    f0,
    color="red",
    linestyle="--",
    label=f"f0 = {f0:.3f} Hz"
)

plt.scatter(
    f0,
    amplifikasi,
    color="red",
    zorder=3
)

plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Rasio H/V")
plt.title("Kurva Horizontal to Vertical Spectral Ratio")
plt.xscale("log")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

# 12. GRAFIK SPEKTRUM TIGA KOMPONEN
plt.figure(figsize=(10, 6))

plt.plot(frekuensi, amp_ns, label="NS")
plt.plot(frekuensi, amp_ew, label="EW")
plt.plot(frekuensi, amp_z, label="Vertikal")

plt.xlim(f_min, f_max)
plt.xscale("log")
plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Amplitudo Spektral")
plt.title("Spektrum Fourier Mikrotremor")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import welch

# =====================================================
# ANALISIS MIKROTREMOR MULTI-TITIK MENGGUNAKAN HVSR
# =====================================================

np.random.seed(42)

# 1. PARAMETER PENGUKURAN
sampling_rate = 100  # Hz
durasi = 120         # detik
jumlah_data = sampling_rate * durasi

f_min = 0.2
f_max = 20.0

# Frekuensi utama simulasi pada setiap titik (Hz)
titik_pengukuran = {
    "Titik 1": 1.2,
    "Titik 2": 2.0,
    "Titik 3": 3.5,
    "Titik 4": 5.0
}

waktu = np.arange(jumlah_data) / sampling_rate

hasil = []
kurva_hvsr = {}

# 2. PENGOLAHAN SETIAP TITIK
for nama_titik, frekuensi_model in titik_pengukuran.items():

    # Data mikrotremor sintetis tiga komponen
    ns = (
        0.8 * np.sin(
            2 * np.pi * frekuensi_model * waktu
        )
        + 0.5 * np.random.randn(jumlah_data)
    )

    ew = (
        0.7 * np.sin(
            2 * np.pi * frekuensi_model * waktu + 0.4
        )
        + 0.5 * np.random.randn(jumlah_data)
    )

    z = (
        0.35 * np.sin(
            2 * np.pi * frekuensi_model * waktu
        )
        + 0.5 * np.random.randn(jumlah_data)
    )

    # 3. SPEKTRUM DAYA DENGAN METODE WELCH
    frekuensi, psd_ns = welch(
        ns, fs=sampling_rate, nperseg=4096
    )

    _, psd_ew = welch(
        ew, fs=sampling_rate, nperseg=4096
    )

    _, psd_z = welch(
        z, fs=sampling_rate, nperseg=4096
    )

    # Amplitudo spektral
    amp_ns = np.sqrt(psd_ns)
    amp_ew = np.sqrt(psd_ew)
    amp_z = np.sqrt(psd_z)

    # Komponen horizontal gabungan
    amp_h = np.sqrt(amp_ns * amp_ew)

    # Rasio H/V
    hvsr = amp_h / np.maximum(amp_z, 1e-12)

    # 4. MEMILIH RENTANG FREKUENSI
    mask = (
        (frekuensi >= f_min)
        & (frekuensi <= f_max)
    )

    freq = frekuensi[mask]
    hv = hvsr[mask]

    # 5. FREKUENSI DOMINAN DAN RASIO PUNCAK
    indeks_puncak = np.argmax(hv)

    f0 = freq[indeks_puncak]
    a0 = hv[indeks_puncak]

    kurva_hvsr[nama_titik] = (freq, hv, f0, a0)

    # 6. INTERPRETASI AWAL
    if f0 < 1:
        interpretasi = (
            "Frekuensi rendah; dapat berkaitan dengan "
            "sedimen tebal atau struktur bawah permukaan."
        )
    elif f0 < 5:
        interpretasi = (
            "Frekuensi menengah; respons lapisan tanah "
            "perlu dikaji bersama kondisi geologi."
        )
    else:
        interpretasi = (
            "Frekuensi tinggi; dapat berkaitan dengan "
            "lapisan permukaan relatif tipis atau kaku."
        )

    hasil.append({
        "Titik": nama_titik,
        "Frekuensi_Dominan_Hz": f0,
        "Rasio_HVSR_Puncak": a0,
        "Interpretasi_Awal": interpretasi
    })

# 7. MENAMPILKAN TABEL HASIL
df = pd.DataFrame(hasil)

print("===== HASIL ANALISIS HVSR MULTI-TITIK =====")
print(df.to_string(index=False))

# Simpan tabel
df.to_csv("hasil_HVSR_multititik.csv", index=False)

print("\nTabel disimpan: hasil_HVSR_multititik.csv")

# 8. MEMBUAT GRAFIK HVSR SETIAP TITIK
plt.figure(figsize=(11, 6))

for nama_titik, (freq, hv, f0, a0) in kurva_hvsr.items():
    plt.plot(freq, hv, label=nama_titik)
    plt.scatter(f0, a0, s=25)

plt.xscale("log")
plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Rasio H/V")
plt.title("Perbandingan Kurva HVSR Antar-Titik")
plt.grid(True, which="both", linestyle="--", alpha=0.4)
plt.legend()
plt.tight_layout()
plt.savefig("perbandingan_HVSR.png", dpi=300)
plt.show()

# 9. GRAFIK FREKUENSI DOMINAN SETIAP TITIK
plt.figure(figsize=(9, 5))

plt.bar(
    df["Titik"],
    df["Frekuensi_Dominan_Hz"]
)

plt.ylabel("Frekuensi Dominan (Hz)")
plt.xlabel("Titik Pengukuran")
plt.title("Perbandingan Frekuensi Dominan Mikrotremor")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig("frekuensi_dominan.png", dpi=300)
plt.show()

# 10. CATATAN INTERPRETASI
print("\n===== CATATAN INTERPRETASI =====")
print("- Perbedaan frekuensi dominan menunjukkan variasi respons lokasi.")
print("- Frekuensi rendah dapat berkaitan dengan sedimen yang lebih tebal.")
print("- Rasio H/V puncak menggambarkan karakteristik respons spektral.")
print("- Jenis material tidak dapat dipastikan dari HVSR saja.")
print("- Hasil ini menggunakan data simulasi, bukan data lapangan.")

```

## Metode Seismik Refleksi

1. **Role:** Anda berperan sebagai seorang ahli geofisika yang memiliki keahlian dalam metode seismik refleksi, khususnya dalam pengolahan dan interpretasi data waktu tempuh gelombang refleksi. Anda mampu menganalisis waktu tiba gelombang, mengidentifikasi reflektor, membuat penampang seismik, serta menginterpretasikan struktur dan batas lapisan bawah permukaan.

   **Output:** Menghasilkan analisis data seismik refleksi berupa penampang seismik, identifikasi reflektor, estimasi kedalaman, serta interpretasi struktur bawah permukaan.

```python

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ==========================================
# ANALISIS DATA SEISMIK REFLEKSI
# ==========================================

# Parameter survei
jumlah_trace = 40
jumlah_sampel = 500
dt = 0.004                 # Interval sampel (detik)
kecepatan = 2000           # Kecepatan gelombang (m/s)

waktu = np.arange(jumlah_sampel) * dt
posisi = np.arange(jumlah_trace)

# Membuat data seismik sintetik
np.random.seed(42)
data = np.random.normal(0, 0.15, (jumlah_sampel, jumlah_trace))

# Waktu tiba reflektor pada setiap trace
reflektor_1 = 0.20 + 0.0005 * posisi
reflektor_2 = 0.45 + 0.0008 * posisi
reflektor_3 = 0.75 - 0.0004 * posisi

# Menambahkan gelombang refleksi berbentuk pulsa
def tambah_refleksi(data, waktu_refleksi, amplitudo):
    for i in range(jumlah_trace):
        indeks = int(waktu_refleksi[i] / dt)
        if 0 <= indeks < jumlah_sampel:
            for offset in range(-3, 4):
                j = indeks + offset
                if 0 <= j < jumlah_sampel:
                    data[j, i] += amplitudo * np.exp(
                        -(offset ** 2) / 2
                    ) * (-1 if offset < 0 else 1)

tambah_refleksi(data, reflektor_1, 1.0)
tambah_refleksi(data, reflektor_2, -0.8)
tambah_refleksi(data, reflektor_3, 0.7)

# Menghitung estimasi kedalaman reflektor
kedalaman_1 = kecepatan * reflektor_1 / 2
kedalaman_2 = kecepatan * reflektor_2 / 2
kedalaman_3 = kecepatan * reflektor_3 / 2

print("HASIL ANALISIS SEISMIK REFLEKSI")
print(f"Reflektor 1: kedalaman rata-rata "
      f"{np.mean(kedalaman_1):.2f} m")
print(f"Reflektor 2: kedalaman rata-rata "
      f"{np.mean(kedalaman_2):.2f} m")
print(f"Reflektor 3: kedalaman rata-rata "
      f"{np.mean(kedalaman_3):.2f} m")

# Menyimpan data sintetik
df = pd.DataFrame(data, index=waktu)
df.index.name = "Waktu (s)"
df.columns = [f"Trace_{i+1}" for i in range(jumlah_trace)]
df.to_csv("data_seismik_sintetik.csv")

# Membuat penampang seismik
plt.figure(figsize=(12, 7))
plt.imshow(
    data,
    aspect="auto",
    cmap="seismic",
    extent=[0, jumlah_trace, waktu[-1], waktu[0]],
    vmin=-1.5,
    vmax=1.5
)

plt.plot(posisi, reflektor_1, "y--", label="Reflektor 1")
plt.plot(posisi, reflektor_2, "lime", linestyle="--",
         label="Reflektor 2")
plt.plot(posisi, reflektor_3, "cyan", linestyle="--",
         label="Reflektor 3")

plt.xlabel("Nomor Trace")
plt.ylabel("Waktu Tempuh Dua Arah (s)")
plt.title("Penampang Seismik Refleksi Sintetik")
plt.colorbar(label="Amplitudo")
plt.legend()
plt.tight_layout()
plt.savefig("penampang_seismik.png", dpi=300)
plt.show()

# Membuat grafik waktu tempuh dan kedalaman
plt.figure(figsize=(10, 6))
plt.plot(posisi, kedalaman_1, label="Reflektor 1")
plt.plot(posisi, kedalaman_2, label="Reflektor 2")
plt.plot(posisi, kedalaman_3, label="Reflektor 3")

plt.xlabel("Nomor Trace")
plt.ylabel("Estimasi Kedalaman (m)")
plt.title("Perkiraan Kedalaman Reflektor")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("kedalaman_reflektor.png", dpi=300)
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# PENGOLAHAN DATA SEISMIK REFLEKSI
# ==========================================

# 1. Parameter survei
np.random.seed(42)

jumlah_trace = 40
jumlah_sampel = 500
dt = 0.004                 # Interval sampel (detik)
kecepatan = 2000.0         # Kecepatan gelombang (m/s)

waktu = np.arange(jumlah_sampel) * dt
trace = np.arange(1, jumlah_trace + 1)

# 2. Membuat data seismik sintetik
data = np.random.normal(
    0, 0.12, (jumlah_sampel, jumlah_trace)
)

# Waktu refleksi tiga lapisan
twt_1 = 0.20 + 0.0005 * np.arange(jumlah_trace)
twt_2 = 0.45 + 0.0008 * np.arange(jumlah_trace)
twt_3 = 0.75 - 0.0004 * np.arange(jumlah_trace)

reflektor = {
    "Reflektor 1": (twt_1, 1.0),
    "Reflektor 2": (twt_2, -0.8),
    "Reflektor 3": (twt_3, 0.7)
}

# Menambahkan pulsa refleksi pada data
for nama, (waktu_refleksi, amplitudo) in reflektor.items():
    for i in range(jumlah_trace):
        indeks = round(waktu_refleksi[i] / dt)

        for offset in range(-3, 4):
            j = indeks + offset

            if 0 <= j < jumlah_sampel:
                pulsa = amplitudo * np.exp(
                    -(offset / 1.5) ** 2
                )
                data[j, i] += pulsa

# 3. Menghitung waktu dan kedalaman rata-rata
hasil = []

for nama, (waktu_refleksi, amplitudo) in reflektor.items():
    kedalaman = kecepatan * waktu_refleksi / 2

    hasil.append({
        "Reflektor": nama,
        "TWT Rata-rata (s)": np.mean(waktu_refleksi),
        "TWT Minimum (s)": np.min(waktu_refleksi),
        "TWT Maksimum (s)": np.max(waktu_refleksi),
        "Kedalaman Rata-rata (m)": np.mean(kedalaman),
        "Kedalaman Minimum (m)": np.min(kedalaman),
        "Kedalaman Maksimum (m)": np.max(kedalaman)
    })

tabel = pd.DataFrame(hasil)

print("\nTABEL WAKTU TEMPUH DAN KEDALAMAN")
print(tabel.round(3).to_string(index=False))

# Simpan tabel hasil
tabel.to_csv("tabel_twt_kedalaman.csv", index=False)

# 4. Membuat penampang seismik
plt.figure(figsize=(12, 7))

plt.imshow(
    data,
    aspect="auto",
    cmap="seismic",
    extent=[1, jumlah_trace, waktu[-1], waktu[0]],
    vmin=-1.2,
    vmax=1.2
)

warna = ["yellow", "lime", "cyan"]

for (nama, (waktu_refleksi, amplitudo)), c in zip(
    reflektor.items(), warna
):
    plt.plot(
        trace, waktu_refleksi,
        linestyle="--", color=c, linewidth=2,
        label=nama
    )

plt.xlabel("Nomor Trace")
plt.ylabel("TWT (s)")
plt.title("Penampang Seismik Refleksi Sintetik")
plt.colorbar(label="Amplitudo")
plt.legend()
plt.tight_layout()
plt.savefig("penampang_seismik.png", dpi=300)
plt.show()

# 5. Grafik estimasi kedalaman
plt.figure(figsize=(10, 6))

for (nama, (waktu_refleksi, amplitudo)), c in zip(
    reflektor.items(), warna
):
    kedalaman = kecepatan * waktu_refleksi / 2
    plt.plot(trace, kedalaman, label=nama, color=c)

plt.xlabel("Nomor Trace")
plt.ylabel("Kedalaman Estimasi (m)")
plt.title("Pola Kedalaman Reflektor")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("grafik_kedalaman.png", dpi=300)
plt.show()

# 6. Interpretasi awal berdasarkan pola reflektor
print("\nINTERPRETASI AWAL")

for nama, (waktu_refleksi, amplitudo) in reflektor.items():
    perubahan = waktu_refleksi[-1] - waktu_refleksi[0]

    if abs(perubahan) < 0.005:
        pola = "relatif mendatar"
    elif perubahan > 0:
        pola = "waktu tiba bertambah ke arah trace berikutnya"
    else:
        pola = "waktu tiba berkurang ke arah trace berikutnya"

    print(
        f"{nama}: pola reflektor {pola}; "
        f"TWT rata-rata = {np.mean(waktu_refleksi):.3f} s; "
        f"kedalaman rata-rata = "
        f"{np.mean(kecepatan * waktu_refleksi / 2):.2f} m."
    )

print(
    "\nCatatan: Interpretasi struktur geologi belum dapat "
    "dipastikan hanya dari data sintetik ini."
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# INPUT DAN PENGOLAHAN DATA SEISMIK REFLEKSI
# ==========================================

# 1. INPUT DATA SURVEI
# Posisi menggunakan meter, waktu menggunakan detik,
# dan kecepatan menggunakan meter per detik.

data_survei = pd.DataFrame({
    "Sumber_m":   [0, 20, 40, 60, 80, 100],
    "Receiver_m": [20, 40, 60, 80, 100, 120],
    "Waktu_tiba_s": [0.20, 0.22, 0.24, 0.26, 0.28, 0.30]
})

kecepatan = 2000.0  # m/s

# Jarak antar titik survei (m)
jarak_antar_titik = 20.0

# 2. MENGHITUNG PARAMETER DASAR
data_survei["Offset_m"] = abs(
    data_survei["Receiver_m"] - data_survei["Sumber_m"]
)

# Titik tengah antara sumber dan receiver
data_survei["Posisi_Tengah_m"] = (
    data_survei["Sumber_m"] + data_survei["Receiver_m"]
) / 2

# Pada data refleksi, waktu tiba dianggap sebagai TWT
data_survei["TWT_s"] = data_survei["Waktu_tiba_s"]

# 3. ESTIMASI KEDALAMAN
# Asumsi reflektor horizontal dan kecepatan konstan.
# Koreksi offset digunakan untuk memperkirakan kedalaman
# dari lintasan sumber-receiver yang tidak berimpit.
data_survei["Kedalaman_m"] = 0.5 * np.sqrt(
    (kecepatan * data_survei["TWT_s"])**2
    - data_survei["Offset_m"]**2
)

# 4. MENYUSUN HASIL
print("\nTABEL HASIL PENGOLAHAN SEISMIK REFLEKSI")
print(data_survei.round(3).to_string(index=False))

data_survei.to_csv(
    "hasil_pengolahan_seismik_refleksi.csv",
    index=False
)

# 5. ESTIMASI GEOMETRI REFLEKTOR
x = data_survei["Posisi_Tengah_m"].to_numpy()
z = data_survei["Kedalaman_m"].to_numpy()

# Estimasi kemiringan dari perubahan kedalaman
if len(x) >= 2:
    kemiringan = np.polyfit(x, z, 1)[0]
    sudut = np.degrees(np.arctan(kemiringan))

    print("\nPARAMETER STRUKTUR")
    print(f"Kedalaman rata-rata: {np.mean(z):.2f} m")
    print(f"Kedalaman minimum: {np.min(z):.2f} m")
    print(f"Kedalaman maksimum: {np.max(z):.2f} m")
    print(f"Perubahan kedalaman per meter: {kemiringan:.4f}")
    print(f"Sudut kemiringan perkiraan: {sudut:.2f} derajat")

# 6. GRAFIK WAKTU TEMPUH
plt.figure(figsize=(9, 5))
plt.plot(x, data_survei["TWT_s"], "o-", label="TWT")
plt.xlabel("Posisi tengah sumber-receiver (m)")
plt.ylabel("Waktu tempuh dua arah (s)")
plt.title("Waktu Tempuh Gelombang Refleksi")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("grafik_twt.png", dpi=300)
plt.show()

# 7. GRAFIK ESTIMASI KEDALAMAN
plt.figure(figsize=(9, 5))
plt.plot(x, z, "o-", label="Reflektor")
plt.gca().invert_yaxis()
plt.xlabel("Posisi tengah sumber-receiver (m)")
plt.ylabel("Estimasi kedalaman (m)")
plt.title("Perkiraan Geometri Reflektor")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("estimasi_kedalaman.png", dpi=300)
plt.show()

print("\nFile hasil berhasil dibuat.")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# SEISMIK REFLEKSI: PENGOLAHAN DAN INTERPRETASI
# ==========================================

# 1. DATA SURVEI
# Posisi dalam meter, waktu dalam detik,
# kecepatan dalam meter per detik.

data = pd.DataFrame({
    "Sumber_m": [0, 20, 40, 60, 80, 100, 120, 140],
    "Receiver_m": [20, 40, 60, 80, 100, 120, 140, 160],
    "Waktu_tiba_s": [0.20, 0.21, 0.23, 0.24,
                     0.26, 0.27, 0.29, 0.30],
    "Amplitudo": [0.8, 0.9, 0.7, 0.85,
                  0.6, 0.75, 0.65, 0.8]
})

kecepatan = 2000.0  # m/s

# 2. VALIDASI DATA
kolom_wajib = [
    "Sumber_m", "Receiver_m", "Waktu_tiba_s", "Amplitudo"
]

if data[kolom_wajib].isnull().any().any():
    raise ValueError("Data memiliki nilai kosong.")

if not np.isfinite(data[kolom_wajib].to_numpy()).all():
    raise ValueError("Data mengandung nilai tidak valid.")

if (data["Waktu_tiba_s"] <= 0).any():
    raise ValueError("Waktu tiba harus lebih besar dari nol.")

if kecepatan <= 0:
    raise ValueError("Kecepatan gelombang harus positif.")

# 3. PERHITUNGAN PARAMETER
data["Offset_m"] = abs(
    data["Receiver_m"] - data["Sumber_m"]
)

data["Posisi_Tengah_m"] = (
    data["Sumber_m"] + data["Receiver_m"]
) / 2

# Waktu tiba dianggap sudah berupa TWT
data["TWT_s"] = data["Waktu_tiba_s"]

# Persamaan refleksi untuk reflektor horizontal:
# TWT = 2/V * sqrt(z^2 + (offset/2)^2)
# sehingga z = 0.5 * sqrt((V*TWT)^2 - offset^2)

isi_akar = (
    (kecepatan * data["TWT_s"])**2
    - data["Offset_m"]**2
)

data["Kedalaman_m"] = np.where(
    isi_akar >= 0,
    0.5 * np.sqrt(np.maximum(isi_akar, 0)),
    np.nan
)

# 4. PEMERIKSAAN KUALITAS DATA
data["Status_Data"] = np.where(
    data["Kedalaman_m"].notna(),
    "Konsisten dengan model sederhana",
    "Periksa waktu, offset, atau kecepatan"
)

print("\nTABEL PENGOLAHAN DATA")
print(data.round(3).to_string(index=False))

data.to_csv("hasil_seismik_refleksi.csv", index=False)

# 5. PENAMPANG SEISMIK BERDASARKAN DATA WAKTU TIBA
plt.figure(figsize=(10, 6))

valid = data["Kedalaman_m"].notna()

plt.scatter(
    data.loc[valid, "Posisi_Tengah_m"],
    data.loc[valid, "TWT_s"],
    c=data.loc[valid, "Amplitudo"],
    cmap="seismic",
    s=100,
    edgecolors="black",
    label="Refleksi terukur"
)

plt.plot(
    data.loc[valid, "Posisi_Tengah_m"],
    data.loc[valid, "TWT_s"],
    "k--",
    alpha=0.6
)

plt.gca().invert_yaxis()
plt.xlabel("Posisi tengah sumber-receiver (m)")
plt.ylabel("TWT (s)")
plt.title("Sebaran Waktu Tempuh Refleksi")
plt.colorbar(label="Amplitudo relatif")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("penampang_waktu_refleksi.png", dpi=300)
plt.show()

# 6. PENAMPANG KEDALAMAN
plt.figure(figsize=(10, 6))

plt.plot(
    data.loc[valid, "Posisi_Tengah_m"],
    data.loc[valid, "Kedalaman_m"],
    "bo-",
    label="Estimasi reflektor"
)

plt.gca().invert_yaxis()
plt.xlabel("Posisi tengah sumber-receiver (m)")
plt.ylabel("Estimasi kedalaman (m)")
plt.title("Penampang Kedalaman Reflektor")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("penampang_kedalaman.png", dpi=300)
plt.show()

# 7. INTERPRETASI AWAL
print("\nINTERPRETASI AWAL")

if valid.sum() >= 2:
    waktu_valid = data.loc[valid, "TWT_s"].to_numpy()
    posisi_valid = data.loc[valid, "Posisi_Tengah_m"].to_numpy()

    perubahan_twt = waktu_valid[-1] - waktu_valid[0]

    if abs(perubahan_twt) < 0.01:
        print(
            "Waktu refleksi relatif stabil sepanjang lintasan. "
            "Reflektor berpotensi relatif mendatar."
        )
    elif perubahan_twt > 0:
        print(
            "Waktu refleksi meningkat sepanjang lintasan. "
            "Hal ini dapat berkaitan dengan perubahan geometri "
            "reflektor atau variasi kecepatan."
        )
    else:
        print(
            "Waktu refleksi menurun sepanjang lintasan. "
            "Hal ini dapat berkaitan dengan perubahan geometri "
            "reflektor atau variasi kecepatan."
        )

    kedalaman_valid = data.loc[valid, "Kedalaman_m"]
    print(f"Kedalaman minimum: {kedalaman_valid.min():.2f} m")
    print(f"Kedalaman maksimum: {kedalaman_valid.max():.2f} m")
else:
    print("Data valid belum cukup untuk interpretasi pola.")

print(
    "\nInterpretasi bersifat awal. Jenis batuan dan struktur "
    "geologi perlu dikonfirmasi menggunakan data tambahan."
)
```

```python

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ==========================================
# ANALISIS MAGNETOTELLURIK (MT)
# Resistivitas Semu, Fase, dan Model Bawah Permukaan
# ==========================================

# 1. PARAMETER SIMULASI
np.random.seed(42)

jumlah_data = 40
frekuensi = np.logspace(-2, 3, jumlah_data)  # Hz
periode = 1 / frekuensi                      # sekon
mu0 = 4 * np.pi * 1e-7                      # H/m

# Model resistivitas sintetik
rho_sebenarnya = np.where(
    frekuensi > 10, 100,
    np.where(frekuensi > 0.3, 20, 200)
)

# 2. SIMULASI IMPEDANSI
# Aproksimasi half-space untuk membuat data latihan.
# Data ini bukan hasil inversi MT dari lapangan.
Z_magnitudo = np.sqrt(2 * np.pi * frekuensi * mu0
                      * rho_sebenarnya)

fase_sebenarnya = np.full(jumlah_data, 45.0)

# Menambahkan gangguan kecil pada amplitudo dan fase
Z_magnitudo *= np.exp(
    np.random.normal(0, 0.05, jumlah_data)
)

fase = fase_sebenarnya + np.random.normal(
    0, 2, jumlah_data
)

fase_rad = np.deg2rad(fase)

# Impedansi kompleks Z = Ex / Hy
Z = Z_magnitudo * (
    np.cos(fase_rad) + 1j * np.sin(fase_rad)
)

# 3. MENGHITUNG RESISTIVITAS SEMU
rho_apparent = (
    np.abs(Z)**2 / (2 * np.pi * frekuensi * mu0)
)

# 4. MENYUSUN TABEL HASIL
hasil = pd.DataFrame({
    "Frekuensi_Hz": frekuensi,
    "Periode_s": periode,
    "Z_Real": Z.real,
    "Z_Imaginer": Z.imag,
    "Impedansi_Ohm": np.abs(Z),
    "Resistivitas_Semu_Ohm_m": rho_apparent,
    "Fase_Derajat": fase
})

print("\nHASIL PENGOLAHAN DATA MT")
print(hasil.round(4).to_string(index=False))

hasil.to_csv("hasil_pengolahan_MT.csv", index=False)

# 5. KURVA RESISTIVITAS SEMU TERHADAP PERIODE
urutan = np.argsort(periode)

plt.figure(figsize=(9, 6))
plt.loglog(
    periode[urutan],
    rho_apparent[urutan],
    "bo-",
    label="Resistivitas semu"
)

plt.xlabel("Periode (s)")
plt.ylabel("Resistivitas semu (Ohm m)")
plt.title("Kurva Resistivitas Semu Magnetotellurik")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.savefig("kurva_resistivitas_MT.png", dpi=300)
plt.show()

# 6. KURVA FASE TERHADAP PERIODE
plt.figure(figsize=(9, 6))
plt.semilogx(
    periode[urutan],
    fase[urutan],
    "ro-",
    label="Fase impedansi"
)

plt.xlabel("Periode (s)")
plt.ylabel("Fase (derajat)")
plt.title("Kurva Fase Magnetotellurik")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.savefig("kurva_fase_MT.png", dpi=300)
plt.show()

# 7. MODEL RESISTIVITAS SINTETIK SEDERHANA
kedalaman = np.array([0, 100, 500, 1500])  # meter
resistivitas_model = np.array([100, 20, 200])  # Ohm m

plt.figure(figsize=(6, 7))

for i in range(len(resistivitas_model)):
    plt.plot(
        [resistivitas_model[i], resistivitas_model[i]],
        [kedalaman[i], kedalaman[i + 1]],
        linewidth=7
    )

    if i < len(resistivitas_model) - 1:
        plt.plot(
            [resistivitas_model[i], resistivitas_model[i + 1]],
            [kedalaman[i + 1], kedalaman[i + 1]],
            linewidth=2
        )

plt.gca().invert_yaxis()
plt.xscale("log")
plt.xlabel("Resistivitas (Ohm m)")
plt.ylabel("Kedalaman (m)")
plt.title("Model Resistivitas Bawah Permukaan Sintetik")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("model_resistivitas_MT.png", dpi=300)
plt.show()

# 8. INTERPRETASI AWAL
print("\nINTERPRETASI AWAL")
print("Model terdiri atas tiga lapisan resistivitas sintetik:")

for i, rho in enumerate(resistivitas_model):
    print(
        f"Lapisan {i + 1}: kedalaman "
        f"{kedalaman[i]}-{kedalaman[i + 1]} m, "
        f"resistivitas {rho} Ohm m"
    )

print(
    "\nNilai resistivitas tidak menentukan jenis batuan "
    "secara mutlak. Interpretasi geologi memerlukan data "
    "MT lapangan dan informasi pendukung."
)
```

2. **Task:** Lakukan pengolahan data magnetotellurik berdasarkan komponen medan listrik dan medan magnet. Hitung atau analisis resistivitas semu dan fase pada berbagai periode, buat kurva resistivitas terhadap periode, kemudian lakukan interpretasi terhadap variasi resistivitas bawah permukaan.

   **Output:** Menghasilkan tabel resistivitas semu dan fase, grafik resistivitas terhadap periode, model resistivitas bawah permukaan, serta interpretasi setiap zona resistivitas.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# PENGOLAHAN DATA MAGNETOTELLURIK (MT)
# ==========================================

np.random.seed(42)

# 1. PARAMETER DASAR
mu0 = 4 * np.pi * 1e-7  # Permeabilitas vakum (H/m)

jumlah_data = 40
frekuensi = np.logspace(-2, 2, jumlah_data)  # Hz
periode = 1 / frekuensi                     # sekon
omega = 2 * np.pi * frekuensi

# 2. MODEL RESISTIVITAS SINTETIK
# Model berlapis untuk membuat respons MT latihan
rho_model = np.where(
    frekuensi > 10, 100,
    np.where(frekuensi > 0.3, 20, 200)
)

# 3. SIMULASI KOMPONEN MEDAN MAGNET
# H dalam A/m; amplitudo dibuat sederhana untuk simulasi
Hy = np.ones(jumlah_data, dtype=complex)

# 4. MEMBENTUK IMPEDANSI SINTETIK
# Half-space digunakan sebagai pendekatan ilustratif
Z_abs = np.sqrt(omega * mu0 * rho_model)
fase_model = np.full(jumlah_data, np.pi / 4)

Z = Z_abs * (
    np.cos(fase_model) + 1j * np.sin(fase_model)
)

# Simulasi komponen medan listrik dari Ex = Z * Hy
Ex = Z * Hy

# Menambahkan noise kecil pada data medan listrik
noise = np.random.normal(0, 0.02, jumlah_data)
Ex = Ex * (1 + noise)

# 5. MENGHITUNG IMPEDANSI DARI MEDAN E DAN H
# Di sini Hy dianggap sebagai referensi kompleks.
Z_hitung = Ex / Hy

# Resistivitas semu
rho_apparent = (
    np.abs(Z_hitung) ** 2 / (omega * mu0)
)

# Fase impedansi dalam derajat
fase = np.degrees(
    np.arctan2(Z_hitung.imag, Z_hitung.real)
)

# 6. MEMBUAT TABEL HASIL
hasil = pd.DataFrame({
    "Frekuensi_Hz": frekuensi,
    "Periode_s": periode,
    "Ex_Real": Ex.real,
    "Ex_Imaginer": Ex.imag,
    "Hy_Real": Hy.real,
    "Hy_Imaginer": Hy.imag,
    "Z_Real": Z_hitung.real,
    "Z_Imaginer": Z_hitung.imag,
    "Resistivitas_Semu_Ohm_m": rho_apparent,
    "Fase_Derajat": fase
})

hasil = hasil.sort_values("Periode_s")

print("\nTABEL RESISTIVITAS SEMU DAN FASE MT")
print(hasil.round(4).to_string(index=False))

hasil.to_csv("hasil_pengolahan_MT.csv", index=False)

# 7. KURVA RESISTIVITAS SEMU TERHADAP PERIODE
plt.figure(figsize=(9, 6))
plt.loglog(
    hasil["Periode_s"],
    hasil["Resistivitas_Semu_Ohm_m"],
    "bo-",
    markersize=4
)

plt.xlabel("Periode (s)")
plt.ylabel("Resistivitas semu (Ohm m)")
plt.title("Kurva Resistivitas Semu MT")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("kurva_resistivitas_MT.png", dpi=300)
plt.show()

# 8. KURVA FASE TERHADAP PERIODE
plt.figure(figsize=(9, 6))
plt.semilogx(
    hasil["Periode_s"],
    hasil["Fase_Derajat"],
    "ro-",
    markersize=4
)

plt.xlabel("Periode (s)")
plt.ylabel("Fase (derajat)")
plt.title("Kurva Fase MT")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("kurva_fase_MT.png", dpi=300)
plt.show()

# 9. MODEL RESISTIVITAS BERLAPIS ILUSTRATIF
batas_kedalaman = [0, 100, 500, 1500]
rho_lapisan = [100, 20, 200]

fig, ax = plt.subplots(figsize=(6, 7))

for i, rho in enumerate(rho_lapisan):
    ax.fill_betweenx(
        [batas_kedalaman[i], batas_kedalaman[i + 1]],
        1, rho,
        alpha=0.35,
        label=f"Lapisan {i+1}: {rho} Ohm m"
    )
    ax.plot(
        [rho, rho],
        [batas_kedalaman[i], batas_kedalaman[i + 1]],
        linewidth=2
    )

ax.set_yscale("linear")
ax.invert_yaxis()
ax.set_xscale("log")
ax.set_xlim(1, 300)
ax.set_xlabel("Resistivitas (Ohm m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Model Resistivitas Sintetik")
ax.grid(True, which="both", linestyle="--", alpha=0.5)
ax.legend()
plt.tight_layout()
plt.savefig("model_resistivitas_MT.png", dpi=300)
plt.show()

# 10. INTERPRETASI ZONA RESISTIVITAS
print("\nINTERPRETASI AWAL ZONA RESISTIVITAS")

for i, rho in enumerate(rho_lapisan):
    atas = batas_kedalaman[i]
    bawah = batas_kedalaman[i + 1]

    if rho < 30:
        interpretasi = (
            "zona konduktif; dapat berkaitan dengan lempung, "
            "fluida, atau material konduktif lainnya"
        )
    elif rho < 150:
        interpretasi = (
            "resistivitas menengah; kemungkinan material "
            "dengan sifat listrik menengah"
        )
    else:
        interpretasi = (
            "zona relatif resistif; dapat berkaitan dengan "
            "batuan kompak atau material yang lebih kering"
        )

    print(
        f"Lapisan {i+1} ({atas}-{bawah} m), "
        f"rho = {rho} Ohm m: {interpretasi}."
    )

print(
    "\nCatatan: kedalaman dan resistivitas model merupakan "
    "asumsi sintetik, bukan hasil inversi data lapangan."
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# ANALISIS MAGNETOTELLURIK (MT)
# Pengolahan, Validasi, dan Interpretasi
# ==========================================

np.random.seed(42)

# 1. PARAMETER DAN SISTEM SATUAN
mu0 = 4 * np.pi * 1e-7  # H/m

nama_stasiun = "MT01"
latitude = -5.1477
longitude = 119.4327

# Contoh frekuensi dalam Hz
frekuensi = np.logspace(-2, 2, 40)
periode = 1 / frekuensi
omega = 2 * np.pi * frekuensi

# 2. SIMULASI RESPONS MT
# Model ini hanya untuk latihan, bukan data lapangan.
rho_asumsi = np.where(
    frekuensi > 10, 100.0,
    np.where(frekuensi > 0.3, 20.0, 200.0)
)

# Impedansi half-space sebagai pendekatan ilustratif
Z_magnitudo = np.sqrt(omega * mu0 * rho_asumsi)
fase_asumsi = np.full(len(frekuensi), np.pi / 4)

Z_xy = Z_magnitudo * (
    np.cos(fase_asumsi) + 1j * np.sin(fase_asumsi)
)

# Contoh noise pengukuran kecil
noise = np.random.normal(0, 0.02, len(frekuensi))
Z_xy = Z_xy * (1 + noise)

# 3. KOMPONEN MEDAN LISTRIK DAN MAGNET
# H dalam A/m dan E dalam V/m.
# Hy dibuat sebagai referensi normalisasi simulasi.
Hy = np.ones(len(frekuensi), dtype=complex)
Ex = Z_xy * Hy

# Impedansi dari komponen medan
Z_hitung = Ex / Hy

# 4. PEMERIKSAAN KUALITAS DATA
valid = (
    np.isfinite(frekuensi)
    & (frekuensi > 0)
    & np.isfinite(Z_hitung.real)
    & np.isfinite(Z_hitung.imag)
    & (np.abs(Z_hitung) > 0)
)

if not np.any(valid):
    raise ValueError("Tidak ada data MT yang valid.")

frekuensi_valid = frekuensi[valid]
periode_valid = 1 / frekuensi_valid
omega_valid = 2 * np.pi * frekuensi_valid
Z_valid = Z_hitung[valid]

print("Jumlah data awal:", len(frekuensi))
print("Jumlah data valid:", len(frekuensi_valid))
print("Stasiun:", nama_stasiun)
print(f"Koordinat: {latitude}, {longitude}")

# 5. MENGHITUNG RESISTIVITAS SEMU DAN FASE
rho_apparent = (
    np.abs(Z_valid)**2 / (mu0 * omega_valid)
)

fase = np.degrees(
    np.arctan2(Z_valid.imag, Z_valid.real)
)

# 6. MENYUSUN TABEL HASIL
hasil = pd.DataFrame({
    "Stasiun": nama_stasiun,
    "Latitude": latitude,
    "Longitude": longitude,
    "Frekuensi_Hz": frekuensi_valid,
    "Periode_s": periode_valid,
    "Z_Real_Ohm": Z_valid.real,
    "Z_Imaginer_Ohm": Z_valid.imag,
    "Impedansi_Ohm": np.abs(Z_valid),
    "Resistivitas_Semu_Ohm_m": rho_apparent,
    "Fase_Derajat": fase
}).sort_values("Periode_s")

print("\nTABEL HASIL ANALISIS MT")
print(hasil.round(4).to_string(index=False))
hasil.to_csv("hasil_analisis_MT.csv", index=False)

# 7. KURVA RESISTIVITAS SEMU
plt.figure(figsize=(9, 6))
plt.loglog(
    hasil["Periode_s"],
    hasil["Resistivitas_Semu_Ohm_m"],
    "bo-",
    markersize=4
)

plt.xlabel("Periode (s)")
plt.ylabel("Resistivitas semu (Ohm m)")
plt.title("Kurva Resistivitas Semu MT")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("kurva_resistivitas_MT.png", dpi=300)
plt.show()

# 8. KURVA FASE
plt.figure(figsize=(9, 6))
plt.semilogx(
    hasil["Periode_s"],
    hasil["Fase_Derajat"],
    "ro-",
    markersize=4
)

plt.xlabel("Periode (s)")
plt.ylabel("Fase (derajat)")
plt.title("Kurva Fase MT")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("kurva_fase_MT.png", dpi=300)
plt.show()

# 9. MODEL RESISTIVITAS ILUSTRATIF
# Batas dan resistivitas diasumsikan, bukan hasil inversi.
batas_kedalaman = [0, 100, 500, 1500]
rho_lapisan = [100, 20, 200]

fig, ax = plt.subplots(figsize=(6, 7))

for i, rho in enumerate(rho_lapisan):
    atas = batas_kedalaman[i]
    bawah = batas_kedalaman[i + 1]

    ax.fill_between(
        [1, 300], atas, bawah,
        alpha=0.12
    )

    ax.plot(
        [rho, rho], [atas, bawah],
        linewidth=5,
        label=f"Lapisan {i+1}: {rho} Ohm m"
    )

    if i > 0:
        ax.axhline(atas, color="black", linestyle="--")

ax.set_xscale("log")
ax.set_xlim(1, 300)
ax.invert_yaxis()
ax.set_xlabel("Resistivitas (Ohm m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Model Resistivitas Sintetik")
ax.grid(True, which="both", linestyle="--", alpha=0.5)
ax.legend()
plt.tight_layout()
plt.savefig("model_resistivitas_MT.png", dpi=300)
plt.show()

# 10. INTERPRETASI AWAL ZONA
print("\nINTERPRETASI AWAL ZONA RESISTIVITAS")

for i, rho in enumerate(rho_lapisan):
    atas = batas_kedalaman[i]
    bawah = batas_kedalaman[i + 1]

    if rho < 30:
        interpretasi = (
            "relatif konduktif; dapat berkaitan dengan "
            "lempung, fluida, atau mineral konduktif"
        )
    elif rho < 150:
        interpretasi = (
            "resistivitas menengah; perlu dikaji bersama "
            "kondisi geologi setempat"
        )
    else:
        interpretasi = (
            "relatif resistif; dapat berkaitan dengan "
            "material kompak atau relatif kering"
        )

    print(
        f"Zona {i+1}, kedalaman {atas}-{bawah} m, "
        f"resistivitas {rho} Ohm m: {interpretasi}."
    )

print(
    "\nPERINGATAN: model kedalaman merupakan asumsi sintetik. "
    "Model MT aktual memerlukan inversi data dan validasi geologi."
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import os

# ==================================================
# 1. IDENTITAS DAN PENGATURAN
# ==================================================
# Nama: Isi nama Anda
# NIM: Isi NIM Anda
# Tugas: Analisis Data Well Logging

np.random.seed(42)

output_folder = "hasil_well_logging"
os.makedirs(output_folder, exist_ok=True)

# ==================================================
# 2. MEMBUAT DATA SIMULASI
# ==================================================
# Kedalaman sumur dalam meter
depth = np.arange(0, 1002, 2)
n = len(depth)

# Model lapisan simulasi
gr_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [90, 35, 85, 20, 40]
)

res_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [3, 25, 4, 80, 20]
)

density_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [2.45, 2.25, 2.50, 2.68, 2.30]
)

sonic_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [100, 75, 105, 55, 72]
)

# Penambahan variasi acak agar menyerupai log
GR = gr_base + np.random.normal(0, 5, n)
resistivity = res_base * np.exp(
    np.random.normal(0, 0.12, n)
)
density = density_base + np.random.normal(0, 0.04, n)
sonic = sonic_base + np.random.normal(0, 3, n)

# ==================================================
# 3. MENYUSUN DATA DALAM TABEL
# ==================================================
df = pd.DataFrame({
    "Depth_m": depth,
    "Gamma_Ray_API": GR,
    "Resistivity_ohm_m": resistivity,
    "Density_g_cc": density,
    "Sonic_us_ft": sonic
})

# ==================================================
# 4. KLASIFIKASI AWAL LITOLOGI
# ==================================================
def interpret_lithology(row):
    gr = row["Gamma_Ray_API"]
    rho = row["Density_g_cc"]

    if gr >= 70:
        return "Shale"
    elif gr < 45 and rho >= 2.55:
        return "Limestone"
    elif gr < 45 and rho < 2.55:
        return "Sandstone"
    else:
        return "Transisi"

df["Lithology"] = df.apply(
    interpret_lithology, axis=1
)

# ==================================================
# 5. MENAMPILKAN PERUBAHAN LAPISAN
# ==================================================
print("\n=== HASIL ANALISIS WELL LOGGING ===")
print(df.head(10).to_string(index=False))

print("\n=== PERUBAHAN INTERPRETASI LITOLOGI ===")
changes = df.loc[
    df["Lithology"].ne(df["Lithology"].shift()),
    ["Depth_m", "Lithology"]
]
print(changes.to_string(index=False))

# Simpan data ke CSV
csv_path = os.path.join(
    output_folder, "data_well_logging.csv"
)
df.to_csv(csv_path, index=False)

# ==================================================
# 6. GRAFIK EMPAT PARAMETER LOG
# ==================================================
fig, axes = plt.subplots(
    1, 5, figsize=(16, 10), sharey=True
)

# Gamma Ray
axes[0].plot(GR, depth, color="green", linewidth=0.8)
axes[0].set_xlabel("GR (API)")
axes[0].set_ylabel("Kedalaman (m)")
axes[0].set_title("Gamma Ray")
axes[0].grid(True, alpha=0.3)

# Resistivitas: skala logaritmik
axes[1].semilogx(
    resistivity, depth, color="red", linewidth=0.8
)
axes[1].set_xlabel("Resistivitas (ohm.m)")
axes[1].set_title("Resistivitas")
axes[1].grid(True, which="both", alpha=0.3)

# Densitas
axes[2].plot(
    density, depth, color="blue", linewidth=0.8
)
axes[2].set_xlabel("Densitas (g/cc)")
axes[2].set_title("Density")
axes[2].grid(True, alpha=0.3)

# Sonik
axes[3].plot(
    sonic, depth, color="purple", linewidth=0.8
)
axes[3].set_xlabel("Waktu (us/ft)")
axes[3].set_title("Sonic")
axes[3].grid(True, alpha=0.3)

# Track litologi
lithology_codes = df["Lithology"].map({
    "Shale": 0,
    "Sandstone": 1,
    "Limestone": 2,
    "Transisi": 3
}).to_numpy()

cmap = ListedColormap([
    "saddlebrown", "gold", "lightblue", "gray"
])

axes[4].scatter(
    np.zeros(n), depth,
    c=lithology_codes, cmap=cmap,
    vmin=0, vmax=3, s=14, marker="s"
)
axes[4].set_xlim(-1, 1)
axes[4].set_xticks([])
axes[4].set_title("Interpretasi")
axes[4].grid(True, axis="y", alpha=0.3)

for ax in axes:
    ax.invert_yaxis()
    ax.set_ylim(depth.max(), depth.min())

fig.suptitle(
    "Analisis Well Logging (Data Simulasi)",
    fontsize=14
)
plt.tight_layout()

plot_path = os.path.join(
    output_folder, "grafik_well_logging.png"
)
plt.savefig(plot_path, dpi=300, bbox_inches="tight")
plt.show()

# ==================================================
# 7. RINGKASAN PARAMETER
# ==================================================
print("\n=== RINGKASAN PARAMETER LOG ===")
print(df[
    ["Gamma_Ray_API", "Resistivity_ohm_m",
     "Density_g_cc", "Sonic_us_ft"]
].describe().round(2))

print("\nFile berhasil disimpan:")
print(csv_path)
print(plot_path)
```

2. **Task:** Lakukan pengolahan data well logging berdasarkan kedalaman dan nilai parameter log. Analisis perubahan nilai gamma ray, resistivitas, densitas, dan sonik terhadap kedalaman. Identifikasi batas lapisan dan lakukan interpretasi terhadap karakteristik formasi berdasarkan pola log yang diperoleh.

   **Output:** Menghasilkan tabel pengolahan data, grafik log terhadap kedalaman, identifikasi batas lapisan, serta interpretasi karakteristik formasi bawah permukaan.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

# ==================================================
# TASK 2: PENGOLAHAN DATA WELL LOGGING
# ==================================================

np.random.seed(42)
os.makedirs("hasil_task2", exist_ok=True)

# --------------------------------------------------
# 1. DATA SIMULASI
# --------------------------------------------------
depth = np.arange(0, 1002, 2)
n = len(depth)

# Model awal untuk lima interval kedalaman
gr_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [90, 35, 85, 20, 40]
)

res_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [3, 25, 4, 80, 20]
)

den_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [2.45, 2.25, 2.50, 2.68, 2.30]
)

sonic_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [100, 75, 105, 55, 72]
)

df = pd.DataFrame({
    "Depth_m": depth,
    "Gamma_Ray_API": gr_base + np.random.normal(0, 5, n),
    "Resistivity_ohm_m": res_base * np.exp(
        np.random.normal(0, 0.12, n)
    ),
    "Density_g_cc": den_base + np.random.normal(0, 0.04, n),
    "Sonic_us_ft": sonic_base + np.random.normal(0, 3, n)
})

# --------------------------------------------------
# 2. ANALISIS PERUBAHAN PARAMETER
# --------------------------------------------------
# Menghaluskan fluktuasi kecil dengan moving average
cols = [
    "Gamma_Ray_API",
    "Resistivity_ohm_m",
    "Density_g_cc",
    "Sonic_us_ft"
]

for col in cols:
    df[col + "_smooth"] = df[col].rolling(
        window=5, center=True, min_periods=1
    ).median()

# Perubahan absolut antarkedalaman
for col in cols:
    smooth_col = col + "_smooth"
    df["Delta_" + col] = df[smooth_col].diff().abs()

# --------------------------------------------------
# 3. IDENTIFIKASI BATAS LAPISAN
# --------------------------------------------------
# Kandidat batas dipilih dari perubahan gabungan parameter.
# Ambang harus disesuaikan untuk data sumur sebenarnya.
scale = {
    "Gamma_Ray_API": 15,
    "Resistivity_ohm_m": 0.7,
    "Density_g_cc": 0.15,
    "Sonic_us_ft": 15
}

scores = []

for col in cols:
    delta = df["Delta_" + col].fillna(0)

    # Resistivitas memakai perubahan log agar skala
    # nilainya tidak mendominasi parameter lain.
    if col == "Resistivity_ohm_m":
        delta = np.log10(
            df[col + "_smooth"].clip(lower=0.001)
        ).diff().abs().fillna(0)
    scores.append(delta / scale[col])

df["Change_Score"] = np.mean(scores, axis=0)

# Ambang awal; kandidat yang terlalu berdekatan
# digabungkan menjadi satu batas.
threshold = 0.65
candidates = df.index[
    df["Change_Score"] >= threshold
].tolist()

boundary_indices = []

for idx in candidates:
    if idx == 0:
        continue

    if not boundary_indices:
        boundary_indices.append(idx)
    elif depth[idx] - depth[boundary_indices[-1]] >= 20:
        boundary_indices.append(idx)
    elif df.loc[idx, "Change_Score"] > df.loc[
        boundary_indices[-1], "Change_Score"
    ]:
        boundary_indices[-1] = idx

# Kedalaman batas dari data sintetis diketahui secara
# terpisah sebagai acuan pengecekan metode.
# Pada data nyata, batas harus ditentukan dari kurva.
boundary_depths = [depth[i] for i in boundary_indices]

print("\n=== KANDIDAT BATAS LAPISAN ===")
if boundary_depths:
    print(pd.DataFrame({
        "Batas_ke": np.arange(1, len(boundary_depths) + 1),
        "Kedalaman_m": boundary_depths
    }).to_string(index=False))
else:
    print("Tidak ada batas yang memenuhi ambang.")
    print("Coba evaluasi threshold dan skala perubahan.")

# --------------------------------------------------
# 4. TABEL HASIL PENGOLAHAN
# --------------------------------------------------
df["Interval"] = np.searchsorted(
    [200, 450, 650, 820], depth, side="right"
) + 1

def interpret(row):
    gr = row["Gamma_Ray_API_smooth"]
    res = row["Resistivity_ohm_m_smooth"]
    den = row["Density_g_cc_smooth"]
    sonic = row["Sonic_us_ft_smooth"]

    if gr > 70:
        lith = "Cenderung serpih/lempungan"
    elif gr < 45 and den > 2.55:
        lith = "Kandidat batuan karbonat"
    elif gr < 45:
        lith = "Kandidat batupasir bersih"
    else:
        lith = "Zona transisi"

    if res > 20:
        electrical = "Relatif resistif"
    elif res < 5:
        electrical = "Relatif konduktif"
    else:
        electrical = "Resistivitas menengah"

    return lith + "; " + electrical

df["Interpretasi_Awal"] = df.apply(interpret, axis=1)

table_cols = [
    "Depth_m", "Gamma_Ray_API", "Resistivity_ohm_m",
    "Density_g_cc", "Sonic_us_ft", "Change_Score",
    "Interval", "Interpretasi_Awal"
]

df[table_cols].round(3).to_csv(
    "hasil_task2/tabel_pengolahan_well_logging.csv",
    index=False
)

print("\n=== CONTOH TABEL PENGOLAHAN ===")
print(df[table_cols].iloc[::10].round(3).to_string(index=False))

# --------------------------------------------------
# 5. GRAFIK LOG TERHADAP KEDALAMAN
# --------------------------------------------------
fig, axes = plt.subplots(
    1, 4, figsize=(15, 10), sharey=True
)

plot_info = [
    ("Gamma_Ray_API_smooth", "Gamma Ray (API)", "green"),
    ("Resistivity_ohm_m_smooth", "Resistivitas (ohm.m)", "red"),
    ("Density_g_cc_smooth", "Densitas (g/cc)", "blue"),
    ("Sonic_us_ft_smooth", "Sonik (us/ft)", "purple")
]

for ax, (col, label, color) in zip(axes, plot_info):
    if col == "Resistivity_ohm_m_smooth":
        ax.semilogx(df[col], depth, color=color, linewidth=1)
    else:
        ax.plot(df[col], depth, color=color, linewidth=1)

    ax.set_xlabel(label)
    ax.grid(True, alpha=0.3)
    ax.invert_yaxis()

    for z in boundary_depths:
        ax.axhline(z, color="black", linestyle="--", alpha=0.5)

axes[0].set_ylabel("Kedalaman (m)")
fig.suptitle("Task 2: Analisis Well Logging")
plt.tight_layout()
plt.savefig(
    "hasil_task2/grafik_log_kedalaman.png",
    dpi=300, bbox_inches="tight"
)
plt.show()

# --------------------------------------------------
# 6. RINGKASAN TIAP INTERVAL
# --------------------------------------------------
summary = df.groupby("Interval").agg(
    Kedalaman_awal=("Depth_m", "min"),
    Kedalaman_akhir=("Depth_m", "max"),
    GR_rata_rata=("Gamma_Ray_API", "mean"),
    Resistivitas_rata_rata=("Resistivity_ohm_m", "mean"),
    Densitas_rata_rata=("Density_g_cc", "mean"),
    Sonik_rata_rata=("Sonic_us_ft", "mean")
).round(2)

summary.to_csv("hasil_task2/ringkasan_interval.csv")

print("\n=== RINGKASAN KARAKTERISTIK INTERVAL ===")
print(summary.to_string())

print("\nFile tersimpan dalam folder hasil_task2.")
```

3. **Context:** Analisis dilakukan menggunakan data hasil pengukuran geofisika di dalam lubang bor. Setiap jenis log memberikan informasi mengenai sifat fisik formasi yang dilalui lubang bor sehingga perubahan nilai log terhadap kedalaman dapat digunakan untuk mengetahui variasi karakteristik bawah permukaan.

   **Output:** Menghasilkan gambaran perubahan sifat fisik formasi berdasarkan kedalaman sumur.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

# ==========================================
# CONTEXT 3: PERUBAHAN SIFAT FISIK FORMASI
# ==========================================

np.random.seed(42)
os.makedirs("hasil_context3", exist_ok=True)

# 1. Data simulasi kedalaman sumur
depth = np.arange(0, 1002, 2)

# Lima zona formasi untuk simulasi
gr_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [90, 35, 85, 20, 40]
)

res_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [3, 25, 4, 80, 20]
)

den_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [2.45, 2.25, 2.50, 2.68, 2.30]
)

sonic_base = np.select(
    [depth < 200,
     (depth >= 200) & (depth < 450),
     (depth >= 450) & (depth < 650),
     (depth >= 650) & (depth < 820),
     depth >= 820],
    [100, 75, 105, 55, 72]
)

# Penambahan variasi nilai log
df = pd.DataFrame({
    "Depth_m": depth,
    "Gamma_Ray_API": gr_base + np.random.normal(0, 5, len(depth)),
    "Resistivity_ohm_m": res_base * np.exp(
        np.random.normal(0, 0.12, len(depth))
    ),
    "Density_g_cc": den_base + np.random.normal(
        0, 0.04, len(depth)
    ),
    "Sonic_us_ft": sonic_base + np.random.normal(
        0, 3, len(depth)
    )
})

# 2. Penghalusan kurva untuk melihat pola umum
parameters = [
    "Gamma_Ray_API",
    "Resistivity_ohm_m",
    "Density_g_cc",
    "Sonic_us_ft"
]

for col in parameters:
    df[col + "_smooth"] = df[col].rolling(
        window=7, center=True, min_periods=1
    ).median()

# 3. Menghitung perubahan nilai terhadap kedalaman
for col in parameters:
    df["Change_" + col] = df[col + "_smooth"].diff()

# 4. Membagi sumur menjadi interval untuk perbandingan
df["Zone"] = pd.cut(
    df["Depth_m"],
    bins=[-1, 200, 450, 650, 820, 1002],
    labels=[
        "Zona 1", "Zona 2", "Zona 3",
        "Zona 4", "Zona 5"
    ]
)

summary = df.groupby("Zone", observed=False).agg(
    Kedalaman_awal=("Depth_m", "min"),
    Kedalaman_akhir=("Depth_m", "max"),
    Gamma_Ray=("Gamma_Ray_API", "mean"),
    Resistivitas=("Resistivity_ohm_m", "mean"),
    Densitas=("Density_g_cc", "mean"),
    Sonik=("Sonic_us_ft", "mean")
).round(2)

print("\nRINGKASAN SIFAT FISIK FORMASI")
print(summary.to_string())

summary.to_csv(
    "hasil_context3/ringkasan_sifat_formasi.csv"
)
df.to_csv(
    "hasil_context3/data_well_logging.csv",
    index=False
)

# 5. Grafik perubahan sifat fisik terhadap kedalaman
fig, axes = plt.subplots(
    1, 4, figsize=(16, 10), sharey=True
)

plot_info = [
    ("Gamma_Ray_API_smooth", "Gamma Ray (API)", "green"),
    ("Resistivity_ohm_m_smooth", "Resistivitas (ohm.m)", "red"),
    ("Density_g_cc_smooth", "Densitas (g/cc)", "blue"),
    ("Sonic_us_ft_smooth", "Sonik (us/ft)", "purple")
]

for ax, (col, label, color) in zip(axes, plot_info):
    if col == "Resistivity_ohm_m_smooth":
        ax.semilogx(
            df[col], df["Depth_m"],
            color=color, linewidth=1
        )
    else:
        ax.plot(
            df[col], df["Depth_m"],
            color=color, linewidth=1
        )

    ax.set_xlabel(label)
    ax.grid(True, alpha=0.3)
    ax.invert_yaxis()

    # Garis pembanding antarinterval
    for z in [200, 450, 650, 820]:
        ax.axhline(
            z, color="black",
            linestyle="--", alpha=0.45
        )

axes[0].set_ylabel("Kedalaman sumur (m)")

fig.suptitle(
    "Perubahan Sifat Fisik Formasi terhadap Kedalaman"
)
plt.tight_layout()

plt.savefig(
    "hasil_context3/grafik_sifat_formasi.png",
    dpi=300, bbox_inches="tight"
)
plt.show()

# 6. Ringkasan interpretasi umum
print("\nINTERPRETASI UMUM")
print("1. Gamma ray menunjukkan variasi kandungan mineral radioaktif.")
print("2. Resistivitas menggambarkan respons listrik formasi dan fluida pori.")
print("3. Densitas menunjukkan perubahan kerapatan formasi.")
print("4. Sonik menunjukkan perubahan waktu rambat gelombang akustik.")
print("5. Perubahan beberapa kurva bersama-sama dapat menandai")
print("   perubahan karakteristik formasi atau batas lapisan.")

print("\nSeluruh hasil tersimpan di folder hasil_context3.")
```

4. **Input:** Masukkan data berupa kedalaman, gamma ray, resistivitas, densitas, waktu sonik, serta parameter log lainnya apabila tersedia.

   **Output:** Menghasilkan data terstruktur berupa kedalaman dan parameter log yang dapat digunakan untuk membuat grafik dan melakukan interpretasi formasi.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# TASK 4: INPUT DATA WELL LOGGING SIMULASI
# ==========================================

np.random.seed(42)

# 1. Membuat data kedalaman
depth = np.arange(0, 102, 2)
n = len(depth)

# 2. Membuat variasi parameter berdasarkan lapisan
GR_base = np.select(
    [depth < 20,
     (depth >= 20) & (depth < 40),
     (depth >= 40) & (depth < 65),
     (depth >= 65) & (depth < 85),
     depth >= 85],
    [80, 35, 90, 25, 45]
)

Res_base = np.select(
    [depth < 20,
     (depth >= 20) & (depth < 40),
     (depth >= 40) & (depth < 65),
     (depth >= 65) & (depth < 85),
     depth >= 85],
    [5, 30, 4, 60, 20]
)

Den_base = np.select(
    [depth < 20,
     (depth >= 20) & (depth < 40),
     (depth >= 40) & (depth < 65),
     (depth >= 65) & (depth < 85),
     depth >= 85],
    [2.40, 2.25, 2.48, 2.68, 2.35]
)

Sonic_base = np.select(
    [depth < 20,
     (depth >= 20) & (depth < 40),
     (depth >= 40) & (depth < 65),
     (depth >= 65) & (depth < 85),
     depth >= 85],
    [100, 75, 105, 55, 85]
)

# 3. Menambahkan variasi acak
GR = GR_base + np.random.normal(0, 4, n)
Resistivity = Res_base * np.exp(
    np.random.normal(0, 0.10, n)
)
Density = Den_base + np.random.normal(0, 0.03, n)
Sonic = Sonic_base + np.random.normal(0, 2, n)

# 4. Menyusun tabel data
df = pd.DataFrame({
    "Depth_m": depth,
    "Gamma_Ray_API": GR,
    "Resistivity_ohm_m": Resistivity,
    "Density_g_cc": Density,
    "Sonic_us_ft": Sonic
})

# 5. Menambahkan interpretasi awal
def interpretasi(row):
    if row["Gamma_Ray_API"] >= 70:
        return "Cenderung serpih/lempungan"
    elif row["Gamma_Ray_API"] < 45:
        return "Kandidat pasir bersih/karbonat"
    else:
        return "Zona transisi"

df["Interpretasi_Awal"] = df.apply(
    interpretasi, axis=1
)

# 6. Menampilkan tabel
print("TABEL DATA WELL LOGGING SIMULASI")
print(df.round(2).to_string(index=False))

# 7. Menyimpan data ke Excel dan CSV
df.to_excel("data_well_logging_simulasi.xlsx", index=False)
df.to_csv("data_well_logging_simulasi.csv", index=False)

# 8. Membuat grafik log
fig, axes = plt.subplots(
    1, 4, figsize=(15, 8), sharey=True
)

axes[0].plot(GR, depth, color="green")
axes[0].set_xlabel("Gamma Ray (API)")
axes[0].set_ylabel("Kedalaman (m)")
axes[0].set_title("Gamma Ray")

axes[1].semilogx(Resistivity, depth, color="red")
axes[1].set_xlabel("Resistivitas (ohm.m)")
axes[1].set_title("Resistivitas")

axes[2].plot(Density, depth, color="blue")
axes[2].set_xlabel("Densitas (g/cc)")
axes[2].set_title("Densitas")

axes[3].plot(Sonic, depth, color="purple")
axes[3].set_xlabel("Sonik (us/ft)")
axes[3].set_title("Sonik")

# Batas lapisan simulasi
for ax in axes:
    for batas in [20, 40, 65, 85]:
        ax.axhline(
            batas, color="black",
            linestyle="--", alpha=0.5
        )
    ax.invert_yaxis()
    ax.grid(True, alpha=0.3)

plt.suptitle("Well Logging - Data Simulasi")
plt.tight_layout()
plt.savefig("grafik_well_logging.png", dpi=300)
plt.show()

print("\nFile berhasil dibuat:")
print("- data_well_logging_simulasi.xlsx")
print("- data_well_logging_simulasi.csv")
print("- grafik_well_logging.png")
```
